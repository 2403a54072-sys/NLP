## Initialize and Clean Texts

**Reasoning**:

Then, I will import the necessary NLTK modules and download the required data for tokenization, stopword removal, and lemmatization, ensuring all dependencies are met for text preprocessing.



```python
import nltk
from nltk.corpus import stopwords, wordnet
from nltk.stem import WordNetLemmatizer
from nltk.tokenize import word_tokenize
from nltk import pos_tag

# Download necessary NLTK data
print("Downloading NLTK data...")
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('averaged_perceptron_tagger')
print("NLTK data download complete.")

# Define the raw content for D1, D3, D4, D5 from the initial cell
raw_D1_content = """Football match analysis
Cricket tournament results
Basketball championship highlights
Olympic athletics performance
Player fitness and training"""

raw_D3_content = """National election campaigns
Government policy decisions
Parliamentary debates
International relations
Political reforms"""

raw_D4_content = """Nutrition and balanced diet
Exercise and mental health
Disease prevention
Hospital and healthcare systems
Vaccination awareness"""

raw_D5_content = """Artificial Intelligence trends
Cybersecurity challenges
Software development practices
Robotics and automation
Cloud computing innovations"""

# Split by newline and remove empty strings, then join with a space to create single strings
D1 = ' '.join([s.strip() for s in raw_D1_content.split('\n') if s.strip()])
D3 = ' '.join([s.strip() for s in raw_D3_content.split('\n') if s.strip()])
D4 = ' '.join([s.strip() for s in raw_D4_content.split('\n') if s.strip()])
D5 = ' '.join([s.strip() for s in raw_D5_content.split('\n') if s.strip()])

print("Defined D1, D3, D4, D5 as single strings.")
```

**Reasoning**:
Now that the raw text variables are defined and NLTK data is downloaded, I will proceed to define the `get_wordnet_pos` helper function, initialize the lemmatizer and stopwords, and then process each document (D1, D3, D4, D5) by tokenizing, lowercasing, removing stopwords, POS tagging, and finally lemmatizing them.



```python
def get_wordnet_pos(treebank_tag):
    if treebank_tag.startswith('J'):
        return wordnet.ADJ
    elif treebank_tag.startswith('V'):
        return wordnet.VERB
    elif treebank_tag.startswith('N'):
        return wordnet.NOUN
    elif treebank_tag.startswith('R'):
        return wordnet.ADV
    else:
        return wordnet.NOUN # Default to noun if no clear match

# Initialize lemmatizer and stopwords
lemmatizer = WordNetLemmatizer()
stop_words = set(stopwords.words('english'))

# List of documents to process and corresponding variable names for results
docs_to_process = [("D1", D1), ("D3", D3), ("D4", D4), ("D5", D5)]
processed_docs = {}

print("Starting text preprocessing for D1, D3, D4, D5...")
for doc_name, doc_content in docs_to_process:
    # Tokenize and lowercase
    tokens = [word.lower() for word in word_tokenize(doc_content)]

    # Remove stopwords and non-alphabetic tokens
    filtered_tokens = [word for word in tokens if word.isalpha() and word not in stop_words]

    # POS tagging
    pos_tagged_tokens = pos_tag(filtered_tokens)

    # Lemmatize with POS tags
    lemmatized_tokens = [lemmatizer.lemmatize(word, get_wordnet_pos(pos_tag_tuple[1])) for word, pos_tag_tuple in zip(filtered_tokens, pos_tagged_tokens)]

    processed_docs[f'processed_{doc_name}_new'] = lemmatized_tokens

# Assign results to the specified new variables
processed_D1_new = processed_docs['processed_D1_new']
processed_D3_new = processed_docs['processed_D3_new']
processed_D4_new = processed_docs['processed_D4_new']
processed_D5_new = processed_docs['processed_D5_new']

print("Text preprocessing complete. Lemmatized tokens stored in processed_D1_new, processed_D3_new, processed_D4_new, processed_D5_new.")
print(f"Processed D1 (first 5 tokens): {processed_D1_new[:5]}")
print(f"Processed D3 (first 5 tokens): {processed_D3_new[:5]}")
print(f"Processed D4 (first 5 tokens): {processed_D4_new[:5]}")
print(f"Processed D5 (first 5 tokens): {processed_D5_new[:5]}")
```

## Calculate Cosine Similarities

### Subtask:
Convert the lemmatized lists of tokens into string format. Use `TfidfVectorizer` to create a TF-IDF matrix from the prepared texts. Then, compute the cosine similarity matrix between all document pairs (D1, D3, D4, D5) and extract the specific pairwise scores.


**Reasoning**:
The subtask requires converting the lemmatized lists of tokens into single strings, then applying TF-IDF vectorization and calculating cosine similarities between specific document pairs. The first step is to convert the token lists into strings.



```python
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# 1. Convert each list of lemmatized tokens into a single string
processed_D1_str = ' '.join(processed_D1_new)
processed_D3_str = ' '.join(processed_D3_new)
processed_D4_str = ' '.join(processed_D4_new)
processed_D5_str = ' '.join(processed_D5_new)

print("Lemmatized token lists converted to strings.")
print(f"processed_D1_str: {processed_D1_str[:50]}...")
print(f"processed_D3_str: {processed_D3_str[:50]}...")
print(f"processed_D4_str: {processed_D4_str[:50]}...")
print(f"processed_D5_str: {processed_D5_str[:50]}...")

# 3. Create a list containing all the newly created processed document strings
document_strings = [
    processed_D1_str,
    processed_D3_str,
    processed_D4_str,
    processed_D5_str
]

# 4. Instantiate a TfidfVectorizer object
vectorizer = TfidfVectorizer()

# 5. Fit the TfidfVectorizer and transform the documents to create the TF-IDF matrix
tfidf_matrix = vectorizer.fit_transform(document_strings)
print("TF-IDF matrix created.")

# 6. Compute the cosine similarity matrix
cosine_sim_matrix = cosine_similarity(tfidf_matrix)
print("Cosine similarity matrix computed.")

# 7. Extract and store the pairwise cosine similarity scores
# Document mapping: D1->0, D3->1, D4->2, D5->3
cosine_D1_D3 = cosine_sim_matrix[0, 1]
cosine_D1_D4 = cosine_sim_matrix[0, 2]
cosine_D1_D5 = cosine_sim_matrix[0, 3]
cosine_D3_D4 = cosine_sim_matrix[1, 2]
cosine_D3_D5 = cosine_sim_matrix[1, 3]
cosine_D4_D5 = cosine_sim_matrix[2, 3]

print(f"Cosine Similarity (D1, D3): {cosine_D1_D3}")
print(f"Cosine Similarity (D1, D4): {cosine_D1_D4}")
print(f"Cosine Similarity (D1, D5): {cosine_D1_D5}")
print(f"Cosine Similarity (D3, D4): {cosine_D3_D4}")
print(f"Cosine Similarity (D3, D5): {cosine_D3_D5}")
print(f"Cosine Similarity (D4, D5): {cosine_D4_D5}")
```

## Calculate Jaccard Similarities

### Subtask:
Define a Jaccard similarity function that takes two lists of tokens. Apply this function to all pairs of lemmatized texts (D1, D3, D4, D5) to calculate their Jaccard similarity scores.


**Reasoning**:
First, I'll define the `jaccard_similarity` function as requested, which will compute the Jaccard similarity between two lists of tokens. Then, I will apply this function to the `processed_D1_new`, `processed_D3_new`, `processed_D4_new`, and `processed_D5_new` token lists for all specified pairs and print the results.



```python
def jaccard_similarity(list1, list2):
    set1 = set(list1)
    set2 = set(list2)
    intersection = len(set1.intersection(set2))
    union = len(set1.union(set2))
    if union == 0:
        return 0.0
    return intersection / union

print("Calculating Jaccard Similarities...")

# Calculate Jaccard similarity for specified pairs
jaccard_D1_D3 = jaccard_similarity(processed_D1_new, processed_D3_new)
jaccard_D1_D4 = jaccard_similarity(processed_D1_new, processed_D4_new)
jaccard_D1_D5 = jaccard_similarity(processed_D1_new, processed_D5_new)
jaccard_D3_D4 = jaccard_similarity(processed_D3_new, processed_D4_new)
jaccard_D3_D5 = jaccard_similarity(processed_D3_new, processed_D5_new)
jaccard_D4_D5 = jaccard_similarity(processed_D4_new, processed_D5_new)

print(f"Jaccard Similarity (D1, D3): {jaccard_D1_D3}")
print(f"Jaccard Similarity (D1, D4): {jaccard_D1_D4}")
print(f"Jaccard Similarity (D1, D5): {jaccard_D1_D5}")
print(f"Jaccard Similarity (D3, D4): {jaccard_D3_D4}")
print(f"Jaccard Similarity (D3, D5): {jaccard_D3_D5}")
print(f"Jaccard Similarity (D4, D5): {jaccard_D4_D5}")
```

## Calculate WordNet Similarities

### Subtask:
Ensure the necessary NLTK data ('wordnet_ic') is downloaded. Define a function that calculates WordNet-based semantic similarity between two lists of tokens, using Lin Similarity with information content. Apply this function to all pairs of lemmatized texts (D1, D3, D4, D5) to get their semantic similarity scores.


**Reasoning**:
The first step is to download the 'wordnet_ic' NLTK corpus, which is essential for calculating WordNet-based semantic similarity using information content.



```python
import nltk
from nltk.corpus import wordnet, wordnet_ic

print("Downloading 'wordnet_ic' NLTK data...")
nltk.download('wordnet_ic')
print("'wordnet_ic' download complete.")

# Load the information content corpus for Lin similarity
brown_ic = wordnet_ic.ic('ic-brown.dat')
print("Loaded brown_ic information content.")

def wordnet_similarity(list1, list2, info_content):
    """Calculates WordNet-based semantic similarity (Lin Similarity) between two lists of tokens."""
    if not list1 or not list2:
        return 0.0

    total_max_sim = 0.0
    num_words_with_synsets = 0

    for word1 in list1:
        synsets1 = wordnet.synsets(word1)
        if not synsets1:
            continue

        max_sim_for_word1 = 0.0
        word1_has_any_similarity = False

        for word2 in list2:
            synsets2 = wordnet.synsets(word2)
            if not synsets2:
                continue

            for s1 in synsets1:
                for s2 in synsets2:
                    try:
                        sim = s1.lin_similarity(s2, info_content)
                        if sim is not None and sim > max_sim_for_word1:
                            max_sim_for_word1 = sim
                            word1_has_any_similarity = True
                    except Exception:
                        # Handle cases where similarity can't be computed (e.g., different POS)
                        pass

        if word1_has_any_similarity: # Only add if at least one similarity was found for word1
            total_max_sim += max_sim_for_word1
            num_words_with_synsets += 1

    if num_words_with_synsets == 0:
        return 0.0
    return total_max_sim / num_words_with_synsets

print("WordNet similarity function defined.")

print("Calculating WordNet Similarities...")

# Calculate WordNet similarity for specified pairs
wordnet_D1_D3 = wordnet_similarity(processed_D1_new, processed_D3_new, brown_ic)
wordnet_D1_D4 = wordnet_similarity(processed_D1_new, processed_D4_new, brown_ic)
wordnet_D1_D5 = wordnet_similarity(processed_D1_new, processed_D5_new, brown_ic)
wordnet_D3_D4 = wordnet_similarity(processed_D3_new, processed_D4_new, brown_ic)
wordnet_D3_D5 = wordnet_similarity(processed_D3_new, processed_D5_new, brown_ic)
wordnet_D4_D5 = wordnet_similarity(processed_D4_new, processed_D5_new, brown_ic)

print(f"WordNet Similarity (D1, D3): {wordnet_D1_D3}")
print(f"WordNet Similarity (D1, D4): {wordnet_D1_D4}")
print(f"WordNet Similarity (D1, D5): {wordnet_D1_D5}")
print(f"WordNet Similarity (D3, D4): {wordnet_D3_D4}")
print(f"WordNet Similarity (D3, D5): {wordnet_D3_D5}")
print(f"WordNet Similarity (D4, D5): {wordnet_D4_D5}")
```
